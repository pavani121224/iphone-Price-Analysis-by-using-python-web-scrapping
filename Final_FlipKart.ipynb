```python
import pandas as pd
import os

df = pd.read_csv("flipkart_products.csv")

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
df.head()

print("\nMissing values:")
print(df.isnull().sum())
```

```python
print(df.to_string(index=False))
print("\nEmpty product names:", (df["Product_Name"].astype(str).str.strip() == "").sum())
print("Empty product prices:", (df["Product_Price"].astype(str).str.strip() == "").sum())
```

```python

df["Price_Numeric"] = (
    df["Product_Price"]
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)

print(df[["Product_Name", "Product_Price", "Price_Numeric"]].head())
```

```python

import pandas as pd
import re

# Extract model name
df["Model"] = df["Product_Name"].str.extract(
    r"(Apple iPhone.*?)(?:\s*\(|$)"
)

# Extract color
df["Color"] = df["Product_Name"].str.extract(
    r"\(([^,]+),"
)

# Extract storage capacity
df["ROM_Storage"] = df["Product_Name"].str.extract(
    r"(\d+\s*(?:GB|TB))\)"
)

# Convert price into numeric format
df["Price_Numeric"] = pd.to_numeric(
    df["Product_Price"]
      .str.replace("₹", "", regex=False)
      .str.replace(",", "", regex=False),
    errors="coerce"
)

print(df[[
    "Product_Name",
    "Model",
    "Color",
    "ROM_Storage",
    "Product_Price",
    "Price_Numeric"
]].head(10).to_string(index=False))
```
