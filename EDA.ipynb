```python
print("=== 1. SUMMARY STATISTICS ===")
print(df_clean[['Price_Numeric', 'Warranty_Duration']].describe())

print("--------------------------------------------------------------------")

# Group-by analysis: Average price by ROM Storage and Model
print("\n=== 2. AVERAGE PRICE BY MODEL & STORAGE ===")
price_summary = df_clean.groupby(['Model', 'ROM_Storage'])['Price_Numeric'].agg(['count', 'mean', 'median', 'std']).reset_index()
print(price_summary)

print("--------------------------------------------------------------------")

# Correlation analysis (if numeric features exist)
numeric_cols = df_clean.select_dtypes(include=[np.number]).columns
if len(numeric_cols) > 1:
    print("\n=== 3. CORRELATION MATRIX ===")
    print(df_clean[numeric_cols].corr())

```

```python
price_series = df_clean['Price_Numeric'].dropna()

adv_stats = pd.Series({
    'Count': price_series.count(),
    'Mean (₹)': price_series.mean(),
    'Median (₹)': price_series.median(),
    'Std Deviation': price_series.std(),
    'Variance': price_series.var(),
    'IQR (Interquartile Range)': price_series.quantile(0.75) - price_series.quantile(0.25),
    'Skewness': price_series.skew(),
    'Kurtosis': stats.kurtosis(price_series)
})
print("=== ADVANCED DESCRIPTIVE STATISTICS ===")
print(adv_stats.round(2))
```

```python
# 3. One-Way ANOVA Test: Testing if mean prices differ significantly across iPhone Models
# Null Hypothesis (H0): All model groups have the same mean price.
# Alternative Hypothesis (Ha): At least one model has a different mean price.
model_groups = [group['Price_Numeric'].dropna().values for name, group in df_clean.groupby('Model')]
f_stat, p_val = stats.f_oneway(*model_groups)

print("\n=== ONE-WAY ANOVA TEST (Price Across Models) ===")
print(f"F-Statistic: {f_stat:.4f}")
print(f"p-value: {p_val:.4e}")
if p_val < 0.05:
    print("Result: Statistically significant difference in prices across models (Reject H0).")
else:
    print("Result: No statistically significant difference in prices across models.")
```

```python
# 4. Chi-Square Test of Independence (Model vs. Storage Capacity)
# Tests whether model choice is independent of storage capacity options
contingency_table = pd.crosstab(df_clean['Model'], df_clean['ROM_Storage'])
chi2, p_chi2, dof, ex = stats.chi2_contingency(contingency_table)

print("\n=== CHI-SQUARE TEST (Model vs. Storage Independence) ===")
print(f"Chi2 Statistic: {chi2:.4f}")
print(f"Degrees of Freedom: {dof}")
print(f"p-value: {p_chi2:.4e}")
```
